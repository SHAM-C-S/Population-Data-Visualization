```python
import matplotlib.pyplot as plt
```

# Bar chart

```python
plt.figure(figsize=(15,6))
plt.bar(papulation["State or union territory"],papulation["Population 2011"])
plt.xlabel("state name")
plt.ylabel("papulation count in 2011")

plt.xticks(rotation=90)
plt.show()
```

# hist plot

```python
plt.figure(figsize=(10, 6))

plt.hist(papulation["Population 2011"], bins=10)

plt.xlabel("Population")
plt.ylabel("Frequency")
plt.title("Population Distribution - 2011")
plt.show()
```
